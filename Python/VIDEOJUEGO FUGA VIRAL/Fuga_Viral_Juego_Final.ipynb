```python
#Este juego ha sido creado de forma equilitativa por las 4 representantes del grupo en clases de taller y tiempo libre
#La introducción fue creada en combinación colaborando mediante GitHub. 
#La prueba 1 fue programada por Ainara Goméz
#La prueba 2 fue programada por Mara García
#La prueba 3 fue programada por Paula Solla
#La prueba 4 y la animación final fue programada por Susana Martin
#Las imagenes utilizadas en el juego fueron creadas utilizando Chatgpt que desarrollo las ilustraciones para aportar un apoyar visual al jugador.

import time #Importamos time para ver el tiempo que nos queda despues de cada jugada
import random #Importamos random para poder elegir preguntas al azar y mezclar sus respuestas
from IPython.display import display, Image #El display muestra contenido del notebook e imagen permite cargar las imagenes
#Con esto conectamos Jupyter con tk (el sistema de ventanas de turtle)
#Asi podemos hacer que la ventana que nos sale responda a acciones del teclado y el temporizador
%gui tk 
import turtle #Importamos la herramienta turtle para poder dibujar las pruebas y en las ventanas que creamos
class FinDelJuego(Exception): #Aqui creamos una excepción. Hereda de exception que es la clase base de excepciones habituales. Lo usamos para detener la celda.
    pass #Dejamos el cuerpo vacio ya que se termina el juego    
# Definimos la función que IPython utilizará para presentar FinDelJuego.
# IPython le proporciona los datos de la excepción mediante estos parámetros.
def finalizar_sin_aviso(shell, tipo, valor, tb, tb_offset=None):
    # Devolvemos una lista vacía ya que no hay líneas de error para mostrar.
    return []
# Obtenemos la sesión de IPython y registramos nuestra función para FinDelJuego.
# La excepción detiene la ejecución sin mostrar el aviso rojo que nos salía antes.
# La coma de (FinDelJuego,) indica que es una tupla de un solo elemento.
get_ipython().set_custom_exc((FinDelJuego,), finalizar_sin_aviso)
# Importamos Output para crear una zona del notebook donde mostrar el texto final.
# Así el mensaje enviado desde Turtle tiene una salida definida.
from ipywidgets import Output

# Guardamos los nombres de las creadoras en una variable de texto.
creadores = "Creado por Susana Martin, Paula Solla, Ainara Gomez y Mara Garcia"
ancho = len(creadores) #Esto nos permite contar los caracteres de la variable "creadores" y lo utilizamos como referencia para la presentación
print("="*ancho)
print("Bienvenidos a:".center(ancho)) #.center nos permite centrar el texto dentro de nuestra variable de ancho
print("FUGA VIRAL".center(ancho))
print("="*ancho)
print(creadores)

print() #Esta linea de print vacio la utilizaremos mucho para separar los prints cuando ejecutemos y que nos visualmente sea mas facil de seguir 

# Repetimos la petición de inicio hasta que el jugador escriba START.
# True mantiene el bucle activo; saldremos de él con break.
while True:
    # .strip elimina los espacios del principio y del final de la respuesta.
    # .lower convierte el texto a minúsculas para aceptar START, Start o start.
    respuesta = input("Escribe START para comenzar").strip().lower()
    if respuesta == "start":
        print("¡Empecemos!")
        break # El break nos permite salir del while y seguir con las instrucciones que hay debajo.
    else: #Sí la respuesta es diferente de "start".
        print("Contesta START o sal del juego")
        
print()

nombre = input("¿Cuál es su nombre, Doctor/a?").strip()

print()
#Utilizamos un f string para sustituir el valor "nombre" por el input del jugador y que se repita durante el juego
print(f"""En el momento más crítico del experimento, un apagón deja al Doctor/a {nombre} atrapado en el laboratorio.
En la oscuridad, el sistema de contención falla y un virus letal comienza a propagarse. 
Está completamente solo y todas las salidas permanecen bloqueadas.
Solo tiene treinta minutos para resolver los enigmas, escapar... y sobrevivir.""")

# Image carga la imagen del archivo indicado en filename.
# width=700 establece su ancho de visualización en 700 píxeles.
# display muestra esa imagen en la salida del notebook.
display(Image(filename=("imagen_fugaviral.png"), width=700))

print()
 
while True: 
    respuesta = input("¿Estás preparado?").strip().lower()
    if respuesta == "si" or respuesta == "sí": # or para que acepte si cumple una de las dos comparaciones.
        print("¡Comienza el juego!")
        inicio = time.time() #desde aqui se guarda el inicio del juego y empieza el tiempo
        penalizacion = 0 #iniciamos a 0 los segundos de penalizacion acumulados
        break 
    elif respuesta == "no": #Si no ha respondido "si" ni "sí", comprobamos si ha respondido "no".
        print("\nGracias por jugar a FUGA VIRAL.") #\n añade un salto de línea antes del mensaje.
        print("El laboratorio te espera para una próxima misión.")
        print("Vuelve a ejecutar esta celda cuando quieras comenzar.")
        raise FinDelJuego #Lanzamos nuestra excepción para detener la ejecución de esta celda.
    else: #Si la respuesta no coincide con ninguna opción válida que hemos establecido antes
        print("Contesta si o no")

if respuesta == "si" or respuesta == "sí":
    print()
    #time.time() - inicio calcula los segundos transcurridos desde el comienzo.
    #También restamos los segundos perdidos por penalizaciones.
    #int() elimina la parte decimal para estar con segundos enteros.
    restante = int(1800 - (time.time() - inicio)- penalizacion) 
    minutos = restante // 60 #La división entera entre 60 obtiene los minutos completos
    segundos = restante % 60 #El resto de dividir entre 60 obtiene los segundos que sobran.
    
    if restante <= 0:
        print("¡Se acabó el tiempo! El virus se ha propagado.")
    else:
        # La f inserta los valores de minutos y segundos.
        # :02d muestra cada entero con al menos dos cifras, añadiendo ceros delante si es solo un digito.
        print(f"Tiempo restante: {minutos:02d}:{segundos:02d}")

print()

print("Para restableceer la luz, conteste correctamente la siguiente pregunta de seguridad.")

print()
#Creamos un diccionario para agrupar los datos de cada pregunta
#Cada clave identifica un dato con so valor verdadero correspondiente
pA = {"pregunta": """¿Cómo se llama el tiempo entre el contagio y la aparición de síntomas?""",
      "opciones": ["Periodo de incubación", "Periodo de vacunación", "Periodo de transmisión", "Periodo de recuperación"],
      "correcta": "Periodo de incubación"} #Guardamos el texto correcto, porque su letra puede cambiar al mezclar.
    
pB = {"pregunta": """¿Qué proceso elimina los microorganismos de los instrumentos del laboratorio?""",
      "opciones": ["Refrigeración", "Evaporación", "Esterilización", "Filtración"],
      "correcta": "Esterilización"}
    
pC = {"pregunta": """¿Qué sistema del cuerpo ayuda a combatir las infecciones?""",
      "opciones": ["Sistema digestivo", "Sistema inmunitario", "Sistema óseo", "Sistema nervioso"], 
      "correcta": "Sistema inmunitario"}

pD = {"pregunta": """¿Cómo se llama la cubierta de proteínas que protege el material genético de un virus?""",
      "opciones": ["Cápside", "Pared celular", "Núcleo", "Ribosoma"],
      "correcta": "Cápside"}

pE = {"pregunta": """¿Qué técnica permite obtener millones de copias de un fragmento específico de ADN?""",
      "opciones": ["PCR", "Centrifugación", "Microscopía", "Filtración"],
      "correcta": "PCR"}

pF = {"pregunta": """¿Qué células se especializan en producir y liberar grandes cantidades de anticuerpos?""",
      "opciones": ["Células plasmáticas", "Linfocitos T citotóxicos", "Glóbulos rojos", "Plaquetas"],
      "correcta": "Células plasmáticas"}
    
preguntas_prueba1 = [pA, pB, pC, pD, pE, pF] #creamos una lista con las posibles preguntas

letras = ["a", "b", "c", "d"] #Las cuatro respuestas que pueden salir

pregunta = random.choice(preguntas_prueba1) #Elegimos una pregunta al azar del banco y el .choice devuelve un elemento de la lista

random.shuffle(pregunta["opciones"]) #Se mezclan las opciones de la pregunta elegida. Shuffle nos modifica la posiicon de la correcta
print(pregunta["pregunta"]) 

for i in range(4): #las cuatro posiciones de las opciones 0, 1, 2, y 3
    print(f"{letras[i]}) {pregunta["opciones"][i]}")
    if pregunta["opciones"][i] == pregunta["correcta"]: #Comparamos el texto de esta opción con el texto de la respuesta correcta.
        letra_correcta = letras[i] #Guardamos la letra que corresponde ahora a la respuesta correcta.
                                   #Es necesario buscarla después de haber mezclado las opciones.
    
while True:
    respuesta = input("Tú respuesta: ").strip().lower()
    restante = int(1800 - (time.time() - inicio) - penalizacion)    
    if restante <= 0:
        print("¡Se acabó el tiempo! El virus se ha propagado.")
        raise FinDelJuego      
    if respuesta not in ["a", "b", "c", "d"]:
        print("Elige a, b, c o d.")     
        print()
    elif respuesta == letra_correcta:
        print("¡Correcto! Las luces vuelven a encenderse.")
        minutos = restante // 60
        segundos = restante % 60
        print(f"Tiempo restante: {minutos:02d}:{segundos:02d}")
        print()
        break    
    else:
        penalizacion += 60
        restante = int(1800 - (time.time() - inicio) - penalizacion)
        if restante <= 0:
            print("¡Se acabó el tiempo! El virus se ha propagado.")
            raise FinDelJuego
        minutos = restante // 60
        segundos = restante % 60
        print(f"Incorrecto. Pierdes 60 segundos. Tiempo restante: {minutos:02d}:{segundos:02d}")
        print()
        pregunta = random.choice(preguntas_prueba1)
        random.shuffle(pregunta["opciones"])
        print(pregunta["pregunta"])
        for i in range(4):
            print(f"{letras[i]}) {pregunta["opciones"][i]}")
            if pregunta["opciones"][i] == pregunta["correcta"]:
                letra_correcta = letras[i]
        print()

print(f"""Felicidaes Doctor/a {nombre} has logrado encender con éxito las luces!!!
Pero aún estas en peligro y tienes que salir de la habitación antes de que el virus se propague y te infecte.
Para ello tienes que buscar el código que abre la puerta del laboratorio.
Como ya sabrás, este código se encuentra escondido en tu mesa de trabajo.""")

print()

prueba_2 = (
    {"imagen": "prueba2_A.png", "codigo": "SALIDA"},
    {"imagen": "prueba2_B.png", "codigo": "BRUVIM"},
    {"imagen": "prueba2_C.png", "codigo": "MACAT1"}
)

prueba_elegida = random.choice(prueba_2)

display(Image(filename=prueba_elegida["imagen"], width=700))
 
while True:
    restante = int(1800 - (time.time() - inicio) - penalizacion)    
    if restante <= 0:
        print("¡Se acabó el tiempo! El virus se ha propagado.")
        raise FinDelJuego  
    respuesta_codigo = input("Introduce la contraseña correcta: ").strip()
    restante = int(1800 - (time.time() - inicio) - penalizacion)    
    if restante <= 0:
        print("¡Se acabó el tiempo! El virus se ha propagado.")
        raise FinDelJuego #detiene antes de la felicitación de prueba superada       
    if respuesta_codigo == prueba_elegida["codigo"]:
        print("ACCESO CONCEDIDO.  El ordenador se ha desbloqueado.")
        minutos = restante // 60
        segundos = restante % 60
        print(f"Tiempo restante: {minutos:02d}:{segundos:02d}")
        print()
        break    
    else:
        penalizacion += 60
        restante = int(1800 - (time.time() - inicio) - penalizacion)
        if restante <= 0:
            print("¡Se acabó el tiempo! El virus se ha propagado.")
            raise FinDelJuego
        minutos = restante // 60
        segundos = restante % 60
        print(f"Incorrecto. Pierdes 60 segundos. Tiempo restante: {minutos:02d}:{segundos:02d}")
        print()
        prueba_elegida = random.choice(prueba_2)
        display(Image(filename=prueba_elegida["imagen"], width=700))

print()

print("""Has conseguido entrar en la siguiente sala. Tras el cristal de un frigorífico encuentras un pequeño vial: la cura del virus mortal. 
Es la última esperanza para salvar a la humanidad, pero debes sacarla del laboratorio sin que pierda su eficacia.
A tu lado hay una nevera de transporte de muestras biológicas.
Coloca el vial en su interior y activa el panel de temperatura.
La cura está a 0 °C. Para estabilizarla, debes alcanzar exactamente 37.5 °C.

INSTRUCCIONES:
• Introduce una suma completa utilizando 7, 5, 3 y 0.5 separados por +.
• El comando suma la temperatura hasta llegar a la temperatura deseada.
• Puedes utilizar los valores más de una vez.
• Calcula bien tus movimientos.Si el resultado no es exactamente 37.5 °C, perderás 60 segundos!

El tiempo sigue corriendo. La cura está en tus manos""")

print()

comandos = {"7": 7, "5": 5, "3": 3, "0.5": 0.5} #Relacionamos cada valor escrito como texto con su valor numérico.
                                                #Las claves llevan comillas porque input devuelve texto.
                                                #Los valores no llevan comillas porque necesitamos sumarlos como números.
temperatura = 0 #iniciamos temperatura antes del primer intnento
temperatura_objetivo = 37.5 #Fijamos la suma de temperatura para superar la prueba

display(Image(filename=("prueba_3.png"), width=700))

while True:
    restante = int(1800 - (time.time() - inicio) - penalizacion)    
    if restante <= 0:
        print("¡Se acabó el tiempo! El virus se ha propagado.")
        raise FinDelJuego 
        
    respuesta_nevera = input("Introduce la suma de valores para llegar a la temperatura deseada").strip()
    respuesta_nevera = respuesta_nevera.replace(" ", "").replace(",", ".")  #Con strip quitamos todos los espacios normales, también los que hay entre los valores.
                                                                            #Con replace sustituimos las comas por puntos para aceptar tanto 0,5 como 0.5.
    restante = int(1800 - (time.time() - inicio) - penalizacion)    

    #comorobamos el tiempo despues y antes de responder
    if restante <= 0: 
        print("¡Se acabó el tiempo! El virus se ha propagado.")
        raise FinDelJuego

    if respuesta_nevera.startswith("+"): #Comprobamos si la suma empieza con un signo "+". Esto nos permite aceptar tanto "+7+5" como "7+5".
       respuesta_nevera = respuesta_nevera[1:]
    #Separamos los valores por cada signo. "7+5+0.5" se convierte en ["7", "5", "0.5"].
    partes = respuesta_nevera.split("+")
    #Comprobamos que todos los valores estén permitidos
    #Any devuelve True si encuentra al menos una parte no permitida. Asi rechazamos valores como "6" o letras no aceptadas.
    if any(parte not in comandos for parte in partes):
        print("Formato no válido. Usa solo 7, 5, 3 y 0.5 separados por +.")
        continue #Saltamos lo que queda de esta vuelta y volvemos al inicio del while.
    #Buscamos el valor numérico de cada parte en el diccionario.
    #sum suma todos esos valores y guardamos el resultado en temperatura.
    #Usamos = porque cada intento sustituye el resultado del intento anterior.
    temperatura = sum(comandos[parte] for parte in partes)
    
    #Comparamos la suma obtenida con la temperatura objetivo.
    # == comprueba si ambos valores son iguales.
    if temperatura == temperatura_objetivo:
        print("¡Temperatura óptima! La cura está lista para transportarse.")
        minutos = restante // 60
        segundos = restante % 60
        print(f"Tiempo restante: {minutos:02d}:{segundos:02d}")
        break

    else: #Si la suma es menor o mayor que 37.5.
        penalizacion += 60
        restante = int(1800 - (time.time() - inicio) - penalizacion) 
        
        #{temperatura:g} muestra el número sin ceros decimales innecesarios.
        print(f"Tu suma es de {temperatura:g} °C y no es la correcta. Pierdes 60 segundos.")

        if restante <= 0:
            print("¡Se acabó el tiempo! El virus se ha propagado.")
            raise FinDelJuego
        minutos = restante // 60
        segundos = restante % 60
        print(f"Tiempo restante: {minutos:02d}:{segundos:02d}")
        print("Inténtalo otra vez con una suma completa.")
        
print()

print(f"""Doctor/a {nombre}, la cura está lista para ser transportada. 
Corres hacia la salida de emergencia, pero al empujar la puerta… no se abre.

El apagón ha activado el cierre de seguridad. Detrás de ti,
una alarma rompe el silencio: el virus ha llegado a esta sala.

Junto a la puerta encuentras una cerradura electrónica.
Tendrás que sincronizarla en tres intentos para desbloquearla.

Si acumulas 3 fallos, perderás 5 minutos y tendrás
que empezar esta prueba de nuevo.

La cura está en tus manos, pero todavía no está a salvo.
Tu misión es salvar a la humanidad, ÁNIMO  no queda nada.""")

print()

print("""Pulsa y suelta ESPACIO cuando el marcador esté en la zona marcada para probar las llaves. 
Necesitas tres aciertos para desbloquear la puerta de emergencia y salir del laboratorio""")

print()

pantalla = turtle.Screen() #Creamos la ventana gráfica donde se desarrollará esta prueba.
pantalla.setup(1200, 600) #Establecemos los tamaños de la ventana que queremos
pantalla.title("Fuga Viral - Última prueba") #Título de la pantalla que se verá en la parte superior
pantalla.bgcolor("grey") #Color de fondo de la pantalla

zona = turtle.Turtle() #Creamos un objeto grafico que representara la zona de acierto
zona.shape("square") #Forma de la zona
zona.color("yellow") #Color de la zona
zona.penup() #La funcion de levar el lapiz de dibujar de turtle al desplazarse
zona.shapesize(stretch_wid=8, stretch_len=6) #Los tamaños de la forma, pero son factores de tamaño no pixeles. Es decir, estiramos vertical 8 veces y 6 horizontal.
zona.goto(0, 0) #Colocamos la zona en el centro de la ventana.

marcador = turtle.Turtle() #Creamos el marcador que se moverá.
marcador.shape("square")
marcador.color("black")
marcador.penup()
marcador.shapesize(stretch_wid=6, stretch_len=0.8)
marcador.speed(0) #Hacemos que cada desplazamiento de Turtle sea inmediato. Speed(0) no detiene el marcador si no que desactiva su animación interna.
marcador.goto(-200, 0) #Lo colocamos en su posición inicial del eje de la pantalla

x = -220 #la posicion horizontal que utilizara la funcion mover()
velocidad = 12 #Cuanto cambia x en cada actualizacion. Al ser positivo lo mueve a la derecha, si es negativo vice versa.

aciertos = 0 #Empezamos sin aciertos y sin fallos
fallos = 0
puerta_abierta = False #Guardamos si ya se ha superado la prueba. False significa que la puerta sigue cerrada, cabiará a true cuando se abra

texto_aciertos = turtle.Turtle() #Creamos objeto para mostrar el numero de aciertos
texto_aciertos.hideturtle() #Ocultamos la figura, pero se podrá seguir escribiendo texto

texto_aciertos.penup() #Evitamos que dibuje una línea al colocarlo.
texto_aciertos.goto(0, 220) #Situamos el contador en la parte superior, centrado

def actualizar_aciertos(): #Definimos una función para volver a mostrar el contador actualizado. Definirla no la ejecuta todavía.
    texto_aciertos.clear() #Borramos el texto anterior de este objeto para evitar superposiciones.
    texto_aciertos.write( #Escribimos cuántos aciertos llevamos de los tres necesarios.
    f"Aciertos: {aciertos}/3", #La f permite insertar el valor.
    align = "center", 
    font = ("Arial", 22, "bold")
    )
#Ejecutamos la función por primera vez para mostrar Aciertos: 0/3.
actualizar_aciertos()

texto_final = turtle.Turtle() #Creamos un objeto para los mensajes de victoria, derrota o penalización.
texto_final.hideturtle() #Ocultamos su figura, pero conservamos su capacidad de escribir.
texto_final.penup() #Evitamos que deje una línea al desplazarse.
texto_final.goto(0,-150) #Colocamos los mensajes debajo de la zona de juego.

texto_tiempo = turtle.Turtle() #Creamos otro objeto exclusivamente para mostrar el tiempo restante.
texto_tiempo.hideturtle()
texto_tiempo.penup()
texto_tiempo.goto(0, 180)

#Definimos la función que calcula y muestra el tiempo disponible.
def actualizar_reloj():
    if puerta_abierta: #Si ya hemos abierto la puerta, dejamos de actualizar el reloj.
        return

    restante = max(0, int(1800 - (time.time() - inicio) - penalizacion)) #Max elige el mayor entre 0 y el resultado para no mostrar tiempo negativo.
    minutos = restante // 60
    segundos = restante % 60

    texto_tiempo.clear() #Borramos la lectura anterior del reloj.
    texto_tiempo.write(  #Escribimos el tiempo actualizado.
        f"Tiempo restante: {minutos:02d}:{segundos:02d}",
        align="center",
        font=("Arial", 20, "bold")
    )

    if restante > 0: #Programamos otra actualización solamente si todavía queda tiempo.
        pantalla.ontimer(actualizar_reloj, 1000) # Solicitamos que esta misma función se ejecute dentro de 1000 ms.

#Definimos cómo se actualiza el movimiento del marcador.
def mover():
    global x, velocidad # Indicamos que vamos a modificar las variables x y velocidad que se crearon fuera de esta función.
    if puerta_abierta: #Si ya hemos ganado, terminamos esta llamada sin volver a programar movimiento.
        return 

    restante = 1800 - (time.time() - inicio) - penalizacion #Calculamos el tiempo disponible, incluyendo las penalizaciones.

    if restante <=0: 
        texto_final.clear() #Quitamos cualquier mensaje anterior de este objeto.
        texto_final.write(
            "Tiempo agotado, el virus se ha propagado. Fin de juego.",
            align = "center", 
            font = ("Arial", 20, "bold")
        )
        return
            
    if fallos >= 3: #Después de tres fallos, dejamos el marcador quieto esperando el reinicio.
        pantalla.ontimer(mover, 20)  #Seguimos revisando el estado y el tiempo aproximadamente cada 20 ms.
        return #Salimos antes de cambiar la posición.
    #Actualizamos la posición horizontal.
    #Equivale a escribir: x = x + velocidad.     
    x += velocidad
    #Comprobamos si el marcador ha alcanzado alguno de los extremos.
    #or significa que basta con cumplir una de las dos condiciones.
    if x >= 220 or x <= -220:
        velocidad = -velocidad  #Cambiamos el signo para invertir el sentido del movimiento.

    marcador.goto(x, 0) #Colocamos el marcador en la nueva posición, manteniendo su altura.
    pantalla.ontimer(mover,20) #Programamos la siguiente actualización dentro de unos 20 milisegundos.

#Definimos cómo reiniciar la prueba de la puerta después de tres fallos.
def reiniciar_prueba():
    global aciertos, fallos, x, velocidad #Vamos a modificar estas variables que están definidas fuera de la función.

    aciertos = 0 #Borramos los aciertos y los fallos de la ronda anterior.
    fallos = 0 
    x = -220 #Recuperamos la posición inicial de la ronda y el desplazamiento inicial hacia la derecha
    velocidad = 12

    texto_final.clear() #Borramos el aviso de los tres fallos.
    actualizar_aciertos() #Mostramos otra vez Aciertos: 0/3.
    marcador.goto(x, 0)

def comprobar(): #Definimos la función que comprueba el intento del jugador.
    global aciertos, puerta_abierta, x, velocidad, fallos, penalizacion #Indicamos qué variables externas vamos a modificar.
    pantalla.title("PRUEBA FINAL")
    
    if puerta_abierta: #Si la puerta ya está abierta, ignoramos nuevos intentos.
        return

    restante = 1800 - (time.time() - inicio) - penalizacion #Calculamos cuánto tiempo queda en el momento del intento.

    if restante <= 0:  #Comprobamos el tiempo antes de permitir que el jugador sume un acierto.
        texto_final.clear() #Borramos el mensaje anterior.
        texto_final.write(
            "Tiempo agotado, el virus se ha propagado, fin del juego", 
            align = "center", 
            font = ("Arial", 20, "bold")
        )
        return #Impedimos que esta llamada llegue a las condiciones de acierto y victoria.
    
    if fallos >=3: #Si la ronda anterior acabó con tres fallos, esta pulsación inicia otra.
        reiniciar_prueba()
        return #La pulsación usada para reiniciar no cuenta también como un intento.

    if x>= -60 and x <= 60: # Comprobamos si el centro del marcador está dentro de la zona permitida. And exige cumplir ambos límites; -60 y 60 también cuentan como acierto.
        aciertos += 1 #Añadimos un acierto al contador.
        actualizar_aciertos() #Actualizamos el contador visible en la ventana.
        print(f"¡Correcto! Aciertos: {aciertos}/3") #Mostramos el número de aciertos usando llaves dentro de la f-string.

    else: #Entramos aquí cuando el marcador está fuera de la zona.
        fallos += 1  #Añadimos un fallo al contador.
        print("Fuera de la zona, inténtelo otra vez.")  #Mostramos este aviso solamente cuando se ha fallado.
        
    if fallos >= 3: #Comprobamos si hemos alcanzado el límite de tres fallos.
            penalizacion += 300  
            texto_final.write( #Mostramos el aviso de penalización y las instrucciones para volver a intentarlo.
                "Has perdido, has cometido 3 fallos y has perdido 5 minutos de tiempo. \nPulsa ESPACIO para volver a intentar si queda tiempo.",
                align="center",
                font=("Arial",  20, "bold")
            )   
            return #Terminamos este intento; la siguiente pulsación podrá reiniciar si queda tiempo.
        
        
    if aciertos == 3: #Comprobamos si hemos conseguido los tres aciertos necesarios.
        puerta_abierta = True #Registramos la victoria para detener el movimiento e ignorar más intentos.
        texto_final.write(
            f"¡Enhorabuena Doctor/a {nombre}! Has conseguido salir del laboratorio a tiempo de la propagación de virus", #el f" string hace que el valor de nombre que asignamos al princpio se repita aqui
            align="center",
            font=("Arial", 20, "bold")
        )
        
        #Mostrar la animación final después de 2 segundos
        pantalla.ontimer(animacion_final, 2000)
        return

    else: #Si todavía no hemos conseguido tres aciertos, preparamos otro intento.
        print("La puerta sigue cerrada, el tiempo se va acabando, sigue intentando")

        #abs elimina el signo negativo, si lo había.
        #Sumamos 4 para aumentar el desplazamiento por actualización.
        #Así el siguiente intento empieza hacia la derecha y resulta más rápido.
        x = -220
        velocidad = abs(velocidad) + 4
        marcador.goto(x, 0)

#Definimos la función que presenta la animación de final del juego.
#append_stdout escribe directamente en salida_final.
def animacion_final():
    #Primero comprobamos que quede tiempo. Después evaluamos la posición del marcador.
    #Cuando conseguimos tres aciertos, registramos la victoria y programamos el desenlace.
    #Definir la función del final no hace que se ejecute por sí sola
    salida_final.append_stdout(f"""La cerradura se abre con un último clic. 

    Doctor/a {nombre}, atraviesas la puerta del laboratorio
    con la cura a salvo. Detrás de ti, el sistema de emergencia
    sella el edificio y contiene el virus mortal.

    Envías la fórmula al resto del mundo. Los equipos médicos
    comienzan a producir y distribuir la cura.

    Las alarmas se apagan. Los primeros pacientes se recuperan.
    Lo que parecía el fin… se convierte en un nuevo comienzo.

    HAS COMPLETADO FUGA VIRAL.
    HAS SALVADO A LA HUMANIDAD.
    """)
    marcador.hideturtle() #Ocultamos el marcador porque la prueba de la puerta ya ha terminado.
    zona.hideturtle() #Ocultamos también la zona amarilla de acierto.
    texto_aciertos.clear() #Borramos el contador de aciertos de la pantalla
    texto_final.clear() #Borramos el mensaje anterior de la prueba.
    texto_tiempo.clear() #Borramos el reloj para dejar espacio a la animación.

    pantalla.bgcolor("#0e2152")
    pantalla.title("Fuga Viral — Humanidad salvada")

    #Creamos un objeto que dibujará el planeta y sus continentes.
    planeta = turtle.Turtle()
    planeta.hideturtle()
    planeta.penup()
    planeta.goto(0, 0)
    planeta.dot(250, "#218bd1") #Dibuja un circulo que se rellena de se color. 
                                #El color fue elegido del color picker de google
    planeta.speed(0)
    planeta.color("#39875b") #Elegimos el verde para el contorno y el relleno de los continentes.

    #Guardamos una lista de continentes.
    #Cada continente contiene los puntos que recorreremos para dibujar su contorno.
    continentes = [
        # América del Norte
        [(-98, 65), (-65, 95), (-35, 75), (-45, 45),
         (-65, 30), (-50, 15), (-80, 25)],

        # América del Sur
        [(-50, 15), (-25, 0), (-20, -25), (-40, -80),
         (-50, -90), (-60, -40), (-65, -10)],

        # Europa y Asia
        [(5, 75), (35, 95), (70, 80), (100, 55),
         (105, 30), (75, 20), (50, 40), (20, 35), (0, 50)],

        # África
        [(8, 35), (35, 30), (50, 0),
         (30, -45), (12, -25), (0, 5)],

        # Australia
        [(65, -50), (90, -42), (100, -62),
         (85, -80), (63, -68)]
    ]
    #Guardar esos puntos todavía no dibuja nada. Los dos bucles siguientes los recorren y unen
    #Recorremos las listas de puntos, una por cada continente.
    #Este for elige un continente, el siguente for recorre los puntos de el continente elegido.
    for continente in continentes:
        planeta.penup() #Levantamos el lápiz para no unir distintos continentes con una línea.
        planeta.goto(continente[0]) #Nos colocamos en el primer punto del continente.
                                    #El índice 0 corresponde al primer elemento de la lista.
        planeta.pendown() #Bajamos el lápiz para dibujar al desplazarnos.
        planeta.begin_fill() #Registramos el contorno de la figura que rellenaremos.

        #Recorremos los puntos desde el segundo hasta el último.
        #[1:] deja fuera el primer punto, porque ya estamos en el.
        for punto in continente[1:]:
            planeta.goto(punto) #Dibujamos una línea desde la posición actual hasta el siguiente punto.

        planeta.goto(continente[0]) #Regresamos al primer punto para cerrar el contorno.
        planeta.end_fill() #Rellenamos la figura con el color verde elegido.
        planeta.penup() #Levantamos el lápiz antes de pasar al siguiente continente.
    luces = turtle.Turtle() #Creamos un objeto independiente para dibujar las luces.
    luces.hideturtle() #Ocultamos su figura para mostrar únicamente los puntos luminosos.
    luces.penup() #Evitamos que dibuje líneas entre una luz y la siguiente.

    #Guardamos las diez posiciones donde aparecerán las luces.
    #Cada pareja indica una posición horizontal y vertical.
    puntos = [
        (-70, 60), (-40, 80), (-60, 20),
        (-30, -20), (-20, -70), (20, 60),
        (50, 40), (30, 0), (70, -20),
        (60, -70)
    ]

    #Definimos una función que dibuja la luz indicada por indice.
    def encender_luz(indice):
        #Comprobamos que exista una posición con ese índice.
        #len(puntos) devuelve cuántos puntos hay: en este caso, 10.
        #Sus índices van del 0 al 9.
        if indice < len(puntos):
            luces.goto(puntos[indice]) #Nos colocamos en la posición correspondiente a esta luz.
            luces.dot(14, "#65ff83") #Dibujamos un punto verde de 14 píxeles de diámetro.

            pantalla.ontimer( #Programamos el encendido de la siguiente luz.
                #lambda crea una pequeña función sin argumentos.
                #Cuando el temporizador la ejecute, llamará a encender_luz
                #pasando el índice siguiente.
                lambda: encender_luz(indice + 1),
                400 #Esperamos aproximadamente 400 milisegundos entre luces.
            )
        else: #Cuando ya hemos dibujado todas las luces y escribimos el texto final
            luces.goto(0, -190) # Nos colocamos debajo del planeta.
            luces.color("white")
            luces.write(
                "La cura llega al mundo. Humanidad salvada.",
                align="center",
                font=("Arial", 20, "bold")
            )
    #Iniciamos la secuencia dibujando la primera luz.
    # Las siguientes se programarán desde la propia función.
    encender_luz(0)

#Estas lineas estan fuera de la animacion final, y son las que ponenn en marcha la prueba de la puerta
#La animación del finalse ejecuta despues cuando comprobar() registra los aciertos con tiempo disponible
salida_final = Output() #Creamos un área de salida para el relato final dentro del notebook.
display(salida_final) #Mostramos esa área, inicialmente vacía.
                      #Más tarde, append_stdout añadirá el relato cuando se active el true de victoria.
pantalla.onkey(comprobar, "space") #Asociamos la función comprobar con soltar la tecla ESPACIO.
pantalla.listen() #Damos el foco a la ventana de Turtle para que reciba las pulsaciones.
actualizar_reloj() # Mostramos el tiempo por primera vez. La propia función programará las siguientes actualizaciones.
mover() # Iniciamos el movimiento del marcador. La propia función programará sus siguientes movimientos.
```
